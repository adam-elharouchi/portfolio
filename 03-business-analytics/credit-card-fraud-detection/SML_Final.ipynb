## Univariate Analysis

### Histogram

```python
import seaborn as sns
import matplotlib.pyplot as plt

numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns

# Determine the number of rows and columns for the subplots
n_cols = 4 # You can adjust this number
n_rows = (len(numeric_cols) + n_cols - 1) // n_cols

fig, axes = plt.subplots(n_rows, n_cols, figsize=(n_cols * 5, n_rows * 4))
axes = axes.flatten() # Flatten the 2D array of axes for easy iteration

for i, col in enumerate(numeric_cols):
    sns.histplot(df[col], kde=True, ax=axes[i])
    axes[i].set_title(f'Distribution of {col}')
    axes[i].set_xlabel('') # Remove x-label to avoid clutter
    axes[i].set_ylabel('') # Remove y-label to avoid clutter

# Hide any unused subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

- General Observations (V-features): The histograms for V1 through V28 (the anonymized principal components) reveal a wide variety of distributions. Many show distinct non-normal patterns, often with multiple peaks, strong skewness, or heavy tails. The KDE plots clearly illustrate these complex shapes, moving far from a typical bell curve.

- For instance, V1, V2, V3, V5, V7, V8, V10, V12, V14, V17, V20, V21, V23, V27, and V28 exhibit considerable skewness (both positive and negative) and very high kurtosis, as evidenced by their numerical values and the long, thin tails or sharp peaks in their KDE plots. This suggests extreme outliers and distributions that are far from normal.
Some V variables like V4, V9, V11, V13, V15, V16, V18, V19, V22, V24, V25, and V26 appear somewhat more symmetric or less extremely skewed, but still largely deviate from a normal distribution, often displaying flatter peaks or slightly skewed tails.
Time: The histogram for Time shows a relatively uniform distribution with two broad peaks, suggesting two main clusters of transaction times. The KDE confirms this bi-modal or spread-out nature. Its slight negative skewness (-0.035) and negative kurtosis (-1.29) indicate a fairly symmetric distribution but with lighter tails than a normal distribution, consistent with the visual plot.

- Amount: The Amount distribution is highly right-skewed, as is very clear from its histogram and KDE plot, which shows a very sharp peak near zero and a long tail extending to much larger values. The skewness value of 16.98 and a massive kurtosis of 845.09 quantitatively confirm this extreme right-skewness and the presence of numerous outliers (very large transaction amounts).

- Class: The Class variable, being binary (0 for non-fraudulent, 1 for fraudulent), is an extremely imbalanced distribution. The histogram (which would essentially show two bars, one very tall for 0 and one very short for 1) and KDE (a tiny bump at 1) would visually confirm this. The extremely high positive skewness (23.99) and kurtosis (573.89) are expected for such an imbalanced classification target.

- In summary, this dataset is characterized by highly non-normal distributions across most of its features. The Amount and Class variables, in particular, show extreme skewness and kurtosis, highlighting the importance of careful preprocessing and model selection, especially due to the significant class imbalance.

### Plot the distribution of 'Class' across different 'Time' interval

```python
import numpy as np

# 1. Create bins for the 'Time' column
# The 'Time' column represents seconds elapsed between the first transaction and the current transaction.
# Max time is 172792 seconds, which is approx 48 hours. Let's create 48 bins for 1-hour intervals.

# Define the number of bins. Since the max time is ~48 hours, 48 bins would represent roughly 1-hour intervals.
n_bins = 48 # This will create bins of approximately 1 hour each, as max time is ~48 hours

df['Time_Bins'] = pd.cut(df['Time'], bins=n_bins, labels=False, include_lowest=True)

# 2. Filter the DataFrame to include only fraudulent transactions
fraudulent_transactions = df[df['Class'] == 1]

# 3. Count the number of fraudulent transactions within each Time_Bins category
fraud_counts_by_time = fraudulent_transactions['Time_Bins'].value_counts().sort_index()

# 4. Create a bar plot to visualize the count of fraudulent transactions per time bin
plt.figure(figsize=(15, 7))
sns.barplot(x=fraud_counts_by_time.index, y=fraud_counts_by_time.values, color='red')

# 5. Add a descriptive title
plt.title('Number of Fraudulent Transactions Across Time Intervals (Approx. 1-hour Bins)', fontsize=16)

# 6. Label the x-axis and y-axis
plt.xlabel('Time Interval (Bin Number)', fontsize=12)
plt.ylabel('Number of Fraudulent Transactions', fontsize=12)

# Optionally, set x-axis ticks to be less crowded if n_bins is large
if n_bins > 20:
    plt.xticks(np.arange(0, n_bins, 5)) # Show a tick every 5 bins

plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

#Baseline Metrics

##Logistic Regression

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# -----------------------
# 1. Define X and y
# -----------------------
X = df.drop('Class', axis=1)
y = df['Class']

# -----------------------
# 2. Train-test split
# -----------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# -----------------------
# 3. Scale features
# -----------------------
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# -----------------------
# 4. Logistic Regression model
# -----------------------
log_reg = LogisticRegression(
    class_weight='balanced',   # helps with fraud class imbalance
    max_iter=1000,
    random_state=42
)

# Train (fit)
log_reg.fit(X_train_scaled, y_train)

# -----------------------
# 5. Predictions & Accuracy
# -----------------------
y_pred = log_reg.predict(X_test_scaled)

accuracy = accuracy_score(y_test, y_pred)

print("Logistic Regression Accuracy:", accuracy)
```

#Oversampling the Data

```python
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import seaborn as sns
import matplotlib.pyplot as plt

# ---------------------------
# 1. Separate features & target
# ---------------------------
X = df.drop('Class', axis=1)
y = df['Class']

# ---------------------------
# 2. Train-test split
# ---------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# ---------------------------
# 3. Scale features BEFORE SMOTE
# ---------------------------
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# ---------------------------
# 4. Apply SMOTE Oversampling
# ---------------------------
sm = SMOTE(random_state=42)
X_train_resampled, y_train_resampled = sm.fit_resample(X_train_scaled, y_train)

# ---------------------------
# 5. Print class counts
# ---------------------------
print("Before Oversampling:")
print(y_train.value_counts())
print("\nAfter Oversampling:")
print(y_train_resampled.value_counts())

# ---------------------------
# 6. Plot frequency BEFORE oversampling
# ---------------------------
plt.figure(figsize=(14,5))

plt.subplot(1, 2, 1)
sns.countplot(x=y_train, palette=["green", "red"])
plt.title("Class Distribution Before Oversampling", fontsize=14)
plt.xlabel("Class (0 = Non-Fraud, 1 = Fraud)")
plt.ylabel("Frequency")

# Add counts on bars
for i, v in enumerate(y_train.value_counts().values):
    plt.text(i, v + max(y_train.value_counts())*0.02, str(v), ha='center')

# ---------------------------
# 7. Plot frequency AFTER oversampling
# ---------------------------
plt.subplot(1, 2, 2)
sns.countplot(x=y_train_resampled, palette=["green", "red"])
plt.title("Class Distribution After SMOTE Oversampling", fontsize=14)
plt.xlabel("Class (0 = Non-Fraud, 1 = Fraud)")
plt.ylabel("Frequency")

# Add counts on bars
for i, v in enumerate(y_train_resampled.value_counts().values):
    plt.text(i, v + max(y_train_resampled.value_counts())*0.02, str(v), ha='center')

plt.tight_layout()
plt.show()
```

##Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# -----------------------------------
# Train Logistic Regression on SMOTE data
# -----------------------------------
log_reg_smote = LogisticRegression(
    class_weight=None,     # no need for balancing now (SMOTE handled it)
    random_state=42
)

log_reg_smote.fit(X_train_resampled, y_train_resampled)

# -----------------------------------
# Predict on the original (unseen) test set
# -----------------------------------
y_pred_smote = log_reg_smote.predict(X_test_scaled)

# -----------------------------------
# Accuracy
# -----------------------------------
accuracy = accuracy_score(y_test, y_pred_smote)
print("Logistic Regression (SMOTE) Accuracy:", accuracy)

# -----------------------------------
# Optional: confusion matrix + metrics
# -----------------------------------
cm = confusion_matrix(y_test, y_pred_smote)
print("\nConfusion Matrix:\n", cm)

print("\nClassification Report:\n")
print(classification_report(y_test, y_pred_smote, digits=4))
```
